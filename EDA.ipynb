```python
import pandas as pd
```

```python
df = pd.read_csv('Base_Dashboard.csv', sep=',')
```

```python
print("--- 1. ESTRUTURA GERAL ---")
print(f"Total de Linhas: {df.shape[0]} (Esperado: 157 candidatos no inner join)")
print(f"Total de Colunas: {df.shape[1]} (Esperado: 18 colunas)")

print("\n--- 2. VERIFICAÇÃO DE COLUNAS DE GAP ---")
colunas_gap = [col for col in df.columns if 'Gap' in col]
if len(colunas_gap) == 0:
    print("✅ Sucesso: Todas as colunas de 'Gap' foram removidas.")
else:
    print(f"❌ Atenção: Ainda existem colunas de Gap na base: {colunas_gap}")

print("\n--- 3. VERIFICAÇÃO DE NULOS (BURACOS NA BASE) ---")
nulos = df.isnull().sum()
if nulos.sum() == 0:
    print("✅ Sucesso: A base não possui valores nulos. O cruzamento foi perfeito.")
else:
    print("❌ Atenção: Valores nulos encontrados nas seguintes colunas:")
    print(nulos[nulos > 0])
```

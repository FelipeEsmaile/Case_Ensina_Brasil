## Agrupar as notas onde o "Registro Ensina" aparece mais de uma vez.


```python
df_copia = df.copy()
```

```python
# Separamos as colunas que vão sofrer o cálculo matemático
colunas_competencias = [
    'Desenvolvimento Contínuo', 
    'Visão Estratégica', 
    'Colaboração e Parceria', 
    'Adaptação e Inovação', 
    'Capacidade de Execução', 
    'Comunicação Empática'
]
```

```python
# Executamos o agrupamento e resetamos o índice
df_copia_agrupado = df_copia.groupby('Registro_Ensina')[colunas_competencias].mean().reset_index()
```

```python
df_copia_agrupado.head(50)
```

```python
# Arredondamos as notas
df_copia_agrupado = df_copia_agrupado.round(2)
```

```python
df_agrupado = df_copia_agrupado
df_agrupado
```

```python
del df_copia_agrupado
```

## UTILIZANDO O MERGE NA TABELA PARA RESGATAR AS COLUNAS NA TABELA

```python
df_copia = df.copy()
```

```python
# Definimos quais colunas de texto (dados do candidato) queremos resgatar
colunas_contexto = [
    'Registro_Ensina', 
    'Turma_Ensina', 
    'Polo_Ensina', 
    'Raça_Ensina', 
    'Gênero_Ensina', 
    'Grau Acadêmico_Ensina',
    'Função_Avaliador(a)',
    'Escola_Avaliador'
]
```

```python
# Vamos na base original e pegamos apenas essas colunas, removendo as duplicatas. 
# Assim, teremos apenas 1 linha com os dados cadastrais para cada Registro.
df_cadastro = df_copia[colunas_contexto].drop_duplicates(subset='Registro_Ensina')
```

## Utilizando o Merge para unir as tabelas

```python
df_auto = pd.read_csv('Arquivo_Tratado_Autoavaliação')
df_terceiros = pd.read_csv('Arquivo_Tratado_Avaliação_Terceiros')
```

```python
# Fazendo o cruzamento perfeito (mantendo apenas quem está nas duas bases) e adicionando sufixos
df_final = pd.merge(
    df_auto, 
    df_terceiros, 
    on='Registro_Ensina', 
    how='inner', 
    suffixes=('_Auto', '_Terc')
)
```

```python
# Transformando o float (2.024) no texto real de categoria ('2024')
df_final['Turma_Ensina'] = (df_final['Turma_Ensina'] * 1000).astype(int).astype(str)

# Arredondando APENAS as colunas de métricas e gaps, protegendo a nossa Turma!
colunas_para_arredondar = [col for col in df_final.columns if '_Auto' in col or '_Terc' in col]
df_final[colunas_para_arredondar] = df_final[colunas_para_arredondar].round(2)
```

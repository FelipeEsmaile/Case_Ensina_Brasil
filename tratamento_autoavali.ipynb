```python
import pandas as pd
```

## Lendo o CSV

```python
df = pd.read_csv('Competências_Autoavaliação.csv', sep=',')
```

```python
df.head()
```

## Adaptação e Inovação

```python
# Criando uma cópia
df_copia = df.copy()
```

```python
# Tiramos o símbolo de '%' usando manipulação de string (.str.replace)
# Convertemos o texto que sobrou para número decimal (.astype(float))
# Dividimos por 100 para ficar na escala de 0 a 1 (ex: 85 vira 0.85)

df_copia['Adaptação e Inovação'] = df_copia['Adaptação e Inovação'].str.replace('%', '').astype(float) / 100
```

```python
df_copia.head()
```

```python
# Atribuo o df de copia ao original
df = df_copia
```

```python
# Deleto o "df_copia"
del df_copia
```

```python
df
```

## Agrupar as notas onde o "Registro Ensina" aparece mais de uma vez.


```python
df_copia = df.copy()
```

```python
# Separamos as colunas que vão sofrer o cálculo matemático
colunas_competencias = [
    'Desenvolvimento Contínuo', 
    'Visão Estratégica', 
    'Colaboração e Parceria', 
    'Adaptação e Inovação', 
    'Capacidade de Execução', 
    'Comunicação Empática'
]
```

```python
# Executamos o agrupamento e resetamos o índice
df_copia_agrupado = df_copia.groupby('Registro_Ensina')[colunas_competencias].mean().reset_index()
```

```python
df_copia_agrupado.head(50)
```

```python
# Arredondamos as notas
df_copia_agrupado = df_copia_agrupado.round(2)
```

```python
df_agrupado = df_copia_agrupado
df_agrupado
```

```python
del df_copia_agrupado
```

## UTILIZANDO O MERGE NA TABELA PARA RESGATAR AS COLUNAS NA TABELA

```python
df_copia = df.copy()
```

```python
# Definimos quais colunas de texto (dados do candidato) queremos resgatar
colunas_contexto = [
    'Registro_Ensina', 
    'Turma_Ensina', 
    'Polo_Ensina', 
    'Raça_Ensina', 
    'Gênero_Ensina', 
    'Grau Acadêmico_Ensina',
    'Função_Avaliador(a)',
    'Escola_Avaliador'
]
```

```python
# Vamos na base original e pegamos apenas essas colunas, removendo as duplicatas. 
# Assim, teremos apenas 1 linha com os dados cadastrais para cada Registro.
df_cadastro = df_copia[colunas_contexto].drop_duplicates(subset='Registro_Ensina')
```

```python
# Fazemos o "Procv" (Merge) juntando as Notas Médias com os Dados Cadastrais
df_copia_final = pd.merge(df_agrupado, df_cadastro, on='Registro_Ensina', how='left')
```

```python
df_copia_final.head()
```

```python
df = df_copia_final
```

```python
del df_copia_final
```

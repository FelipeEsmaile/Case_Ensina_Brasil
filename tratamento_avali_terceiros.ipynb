## Desenvolvimento Contínuo

```python
# Criando uma copia para o df.
df_copia = df
df_copia
```

```python
# Primeiro, criamos uma lista com o nome exato das colunas "micro"
cols_desenv_continuo = [
    'Desenvolvimento Contínuo_Devolutiva', 
    'Desenvolvimento Contínuo_Diálogo', 
    'Desenvolvimento Contínuo_Compartilha', 
    'Desenvolvimento Contínuo_Referência'
]
```

```python
# Criamos a nova coluna "macro".
# O comando .mean(axis=1) tira a média da linha e já dividimos por 4.
df_copia['Desenvolvimento Contínuo'] = df_copia[cols_desenv_continuo].mean(axis=1) / 4
```

```python
# removemos as colunas "micro" antigas para não poluir a base
df_copia = df_copia.drop(columns=cols_desenv_continuo)
```

```python
# Atribuo o df de copia ao original
df = df_copia
```

```python
# Deleto o "df_copia"
del df_copia
```

## Arredondando as casas decimais das competencias

```python
df_copia = df.copy()
df_copia
```

```python
# Agrupando as colunas das competências em uma lista para facilitar
colunas_competencias = [
    'Desenvolvimento Contínuo', 
    'Visão Estratégica', 
    'Colaboração e Parceria', 
    'Adaptação e Inovação', 
    'Capacidade de Execução', 
    'Comunicação Empática'
]

# Aplicando o arredondamento de 2 casas decimais substituindo os valores antigos
df_copia[colunas_competencias] = df_copia[colunas_competencias].round(2)
```

```python
# Atribuindo "df_copia" a "df"
df = df_copia
```
